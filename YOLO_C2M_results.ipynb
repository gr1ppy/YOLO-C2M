```python
from ultralytics import YOLO

model = YOLO('ultralytics/results/v8_neu/weights/best.pt')

results = model.val(
    data='neu-det.yaml',
    imgsz=640,
    split='val',
    project='runs/val',
    name='val_on_test_run'
)

print(results.box.map)    # Print mAP50-95
print(results.box.map50)  # Print mAP50
```
